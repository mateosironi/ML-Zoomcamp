```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df.head().T
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')

categorical_columns = list(df.dtypes[df.dtypes == 'str'].index)

for c in categorical_columns:
    df[c] = df[c].str.lower().str.replace(' ', '_')

df.head().T
```

```python
df.dtypes
```

```python
tc = pd.to_numeric(df.totalcharges, errors='coerce')
tc.isnull().sum()
```

```python
df[tc.isnull()].T
```

```python
df.totalcharges = pd.to_numeric(df.totalcharges, errors = 'coerce')
df.totalcharges = df.totalcharges.fillna(0)
```

```python
df.churn = (df.churn == 'yes').astype(int)
df.churn.value_counts()
```

```python
df.info()
```

```python
# We split the data into train, validation and test using Scikit-Learn
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)

len(df_train), len(df_val), len(df_test)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
df_full_train = df_full_train.reset_index(drop=True)

y_train = df_train.churn.values
y_val = df_val.churn.values
y_test = df_test.churn.values
y_full_train = df_full_train.values

del df_train['churn']
del df_val['churn']
del df_test['churn']
#del df_full_train['churn']
```

Feature Importance: Churn rate and risk ratio

```python
# Lets start comparing the global churn rate with each categorical churn rate
churn_female = df_full_train[df_full_train.gender == 'female'].churn.mean()
churn_male = df_full_train[df_full_train.gender == 'male'].churn.mean()

print(f'churn_female: {churn_female}, diff: {global_churn - churn_female}')
print(f'churn_male: {churn_male}, diff: {global_churn - churn_male}')
```

```python
df_full_train.partner.value_counts()
```

```python
churn_partner = df_full_train[df_full_train.partner == 'yes'].churn.mean()
churn_no_partner = df_full_train[df_full_train.partner == 'no'].churn.mean()

print(f'churn_partner: {churn_partner}, diff: {global_churn - churn_partner}')
print(f'churn_no_partner: {churn_no_partner}, diff: {global_churn - churn_no_partner}')
```

```python
# Risk ratio: group_churn / global_churn (R<1: less likely to churn; R>1: more likely to churn; R=1: as likely to churn as everyone else)
# Instead of doing this operation manually, we will use pandas 'groupby'

from IPython.display import display

for c in categorical:
    print(c)
    df_group = df_full_train.groupby(c).churn.agg(['mean', 'count'])
    df_group['diff'] = df_group['mean'] - global_churn
    df_group['risk'] = df_group['mean'] / global_churn
    display(df_group)
    print()
    print()
```

One-Hot Encoding (OHE): We transform categorical variables into vectors which length is the ammount of possible categories, and which values take 1 and the rest 0. For that, we first transform the dataframe into a dictionary using DictVectorizer from sklearn.feature_extraction. After that, we use fit_transform to transform the dictionary into a matrix where the categorical columns are the binary vectors.

```python
# DictVectorizer
from sklearn.feature_extraction import DictVectorizer

dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)
```

```python
dv.get_feature_names_out()
```

```python
X_train.shape
```

```python
# We must encode the validation set in exactly the same way as the training set.
# We only call 'transform' for validation, to get a matrix with the same columns
val_dict = df_val[categorical + numerical].to_dict(orient='records')
X_val = dv.transform(val_dict)
X_val.shape
```

Logistic Regression: we train the model with linear regression and the sigmoid function, to return a number between 1 and 0 (probability)

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver='lbfgs')
# solver='lbfgs' is the default solver in newer version of sklearn
# for older versions, you need to specify it explicitly
model.fit(X_train, y_train)
```

```python
model.intercept_[0]
```

```python
model.coef_[0].round(3)
```

```python
# Making predictions: model.predict(x) gives hard predictions (0 or 1); model.predict_proba(x) gives soft predictions (between 0 and 1)

y_pred = model.predict_proba(X_val)[:, 1]
churn_decision = (y_pred >= 0.5)

# Accuracy: The fraction of predictions the model got right
(y_val == churn_decision).mean()
```

```python
df_pred = pd.DataFrame()
df_pred['probability'] = y_pred
df_pred['prediction'] = churn_decision.astype(int)
df_pred['actual'] = y_val

df_pred.head()
```

```python
df_pred['correct'] = df_pred.prediction == df_pred.actual
df_pred.head()
```

```python
df_pred.correct.mean()
```

Model Interpretation: we match each weight with its feature name, and then train a smaller model on three features to see clearly how the weights work.

```python
# Lets pair the 45 weights with their corresponding columns using the 'zip' function.
dict(zip(dv.get_feature_names_out(), model.coef_[0].round(3)))
```

```python
# Lets train a smaller model, with only three variables: contract, tenure and monthlycharges
small = ['contract', 'tenure', 'monthlycharges']

dicts_train_small = df_train[small].to_dict(orient='records')
dicts_val_small = df_val[small].to_dict(orient='records')

dv_small = DictVectorizer(sparse=False)
dv_small.fit(dicts_train_small)

dv_small.get_feature_names_out()
```

```python
X_train_small = dv_small.transform(dicts_train_small)

model_small = LogisticRegression(solver='lbfgs')
model_small.fit(X_train_small, y_train)
```

```python
w0 = model_small.intercept_[0]
w0
```

Final Model Training and Testing

```python
# Now we train the model on df_full_train, and validate it on df_test

dicts_full_train = df_full_train[categorical + numerical].to_dict(orient='records')

dv = DictVectorizer(sparse=False)
X_full_train = dv.fit_transform(dicts_full_train)
y_full_train = df_full_train.churn.values
```

```python
model = LogisticRegression(solver='lbfgs')
model.fit(X_full_train, y_full_train)
```

```python
dicts_test = df_test[categorical + numerical].to_dict(orient='records')
X_test = dv.transform(dicts_test)

y_pred = model.predict_proba(X_test)[:, 1]
churn_decision = (y_pred >= 0.5)
(churn_decision == y_test).mean()
```
