```python
df_reduced = df_asia[['model_year', 'vehicle_weight']]
df_reduced.head(7)
```

```python
X = df_reduced.iloc[0:7,:].to_numpy()
X
```

```python
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
print(f'XTX:\n{XTX}')
print(f'XTX_inv:\n{XTX_inv}')
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)
print(f'Sum of weights: {w.sum()}')
```
